## 1D NUMPY

```python
import numpy as np
arr=np.array([10,11,12,13,14,15])
print("1st five Array:", arr[:5])
print("Last five Array:", arr[-5:])
```

## MATRIX

```python
matrix=np.arange(1,13).reshape(3,4)
print("Matrix:")
print(matrix)
print("Second Row:", matrix[1])
print("Last Column:", matrix[:,-1])
```

## DOT PRODUCT

```python
A=np.array([1,2,3,4])
B=np.array([10,20,30,40])
print("Element wise addition:", A+B)
dot=0
for i in range(len(A)):
    dot=dot+A[i]*B[i]
print("Dot Product:", dot)
```
