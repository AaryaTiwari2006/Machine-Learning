## Statestical measure

#### mean, meadian, mode, std dev, veriance

```python
import pandas as pd
data={
    "Name":["A","B","C","D","E","F","G","H"],
    "Marks":[2,4, 7, 1, 8 ,8, 9,5]
}
df = pd.DataFrame(data)
print("dataset", df)
data=df["Marks"]
mean=data.mean()
median=data.median()
mode=data.mode()[0]
varience=data.var()
std=data.std()
print("Mean=",mean)
print("Median=",median)
print("Mode=", mode)
print("Varience=",varience)
print("Std dev=", std)
```

#### Dataset preprocessing, missing, filter, merge

```python
data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Marks": [2, 4, None, 1, 8, 8, 9, 5]
}
df = pd.DataFrame(data)
print("Before processing:")
print(df)
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print("After preprocessing:")
print(df)
```

```python
data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Marks": [2, 4, 7, 1, 8, 8, 9, 5]
}
df = pd.DataFrame(data)
result=df[df["Marks"]>7]
print(result)
```

## liner regresion MSE and RMSE

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
X = np.array([1,5,7,8,9,11,20,21,25,28,30,32,47,40,42,46,50]).reshape(-1, 1)
Y = np.array([4,6,8,10,15,16,17,19,24,28,32,47,40,48,49,55,56])
model = LinearRegression()
model.fit(X, Y)
Y_pred = model.predict(X)
m = model.coef_[0]
c = model.intercept_
mse = mean_squared_error(Y, Y_pred)
rmse = np.sqrt(mse)
print("X values:", X.flatten())
print("Predicted Y:", Y_pred)
print("Slope =", m)
print("Intercept =", c)
print("Linear regression equation:")
print("Y =", m, "X +", c)
print("MSE:", mse)
print("RMSE:", rmse)
plt.scatter(X, Y, label="Actual Data")
plt.plot(X, Y_pred, label="Best Fit Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")
plt.legend()
plt.show()
```
