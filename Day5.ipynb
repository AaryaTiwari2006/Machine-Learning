# Cross validation using KNN model for K-Fold, POut, OneOut

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import (KFold, LeavePOut,LeaveOneOut, cross_val_score)
from sklearn.neighbors import KNeighborsClassifier

X,y=load_iris(return_X_y=True)
model=KNeighborsClassifier (n_neighbors=5)
#K-Fold
kfold=KFold(n_splits=10,shuffle=True,random_state=42)
score_kfold=cross_val_score(model,X,y,cv=kfold)
print("="*50)
print("10-Fold cross validation")
print("Scores:", score_kfold)
print("Mean Accuracy:", score_kfold.mean())
print("-"*50)
#Leave Pout
lpo=LeavePOut(p=2)
scores_lpo=cross_val_score(model,X,y,cv=lpo)
print("LEAVE-P-OUT CROSS VALIDATION")
print("Mean Accuracy:", scores_lpo.mean())
print("-"*50)
#LeaveOneOut
loo=LeaveOneOut()
scores_loo=cross_val_score(model,X,y,cv=loo)
print("LEAVE-ONE-OUT CROSS VALIDATION")
print("Mean Accuracy:", scores_loo.mean())
print("="*50)
```
